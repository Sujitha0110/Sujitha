```python
import numpy as np
```

```python
from sklearn.linear_model import LinearRegression
```

```python
X=np.array([[12], [5], [15], [4]])
y=np.array([78,45,88,39])
model=LinearRegression()
#train the values
model.fit(X,y)
#give new value it predicts with old value
ans=model.predict([[8]])
print(ans)
X1=X.ravel()
np.corrcoef(X1,y)
```

```python
from ast import Attribute
att=np.array([[85],[62],[91],[55]])
y=np.array([78,45,88,39])
model=LinearRegression()
model.fit(att,y)
ans=model.predict([[74]])
print(ans)
X1=X.ravel()
np.corrcoef(X1,y)
```
